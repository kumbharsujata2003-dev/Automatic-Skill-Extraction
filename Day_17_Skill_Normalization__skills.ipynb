```python
import pandas as pd
import re
```

```python
df = pd.read_csv("raw_jobs.csv")

print(df.shape)
```

```python
print(df["skills"].head(10))
```

```python
skills = df["skills"].fillna("").astype(str)

print(skills.head(10))
```

```python
def clean_skill(skill):
    skill = skill.lower()
    skill = re.sub(r"[^a-z0-9+#.\- ]", "", skill)
    skill = re.sub(r"\s+", " ", skill).strip()
    return skill
```

```python
skill_aliases = {
    "python3": "Python",
    "python 3": "Python",
    "python programming": "Python",
    "powerbi": "Power BI",
    "power bi": "Power BI",
    "power-bi": "Power BI",
    "postgres": "PostgreSQL"
}
```

```python
def normalize_skill(skill):
    skill = clean_skill(skill)

    if skill in skill_aliases:
        return skill_aliases[skill]

    return skill.title()
```

```python
test_skills = [
    "Python",
    "Python3",
    "Python 3",
    "Python programming",
    "PowerBI",
    "Power BI",
    "Power-BI",
    "Postgres"
]

for skill in test_skills:
    print(skill, "→", normalize_skill(skill))
```

```python
df["normalized_skills"] = skills.apply(normalize_skill)

print(df[["skills", "normalized_skills"]].head(10))
```

```python
df.to_csv("normalized_skills.csv", index=False)

print("normalized_skills.csv created successfully!")
```
