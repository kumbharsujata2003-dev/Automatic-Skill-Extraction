```python
import spacy
```

```python
nlp = spacy.load("en_core_web_sm")

print("spaCy model loaded successfully")
```

```python
text = """
Python is a programming language.
Microsoft and AWS are widely used technologies.
Google Cloud provides cloud services.
New York is a major city.
"""

doc = nlp(text)

for ent in doc.ents:
    print(ent.text, "->", ent.label_)
```

Standard NER models are trained on general entities such as people,
organizations, locations, and dates. Many technical skills such as
Python, Power BI, and PostgreSQL are domain-specific terms, so they
may not be recognized correctly by a standard NER model.

```python
TRAIN_DATA = [
    (
        "Python is used for data analysis.",
        {"entities": [(0, 6, "SKILL")]}
    ),
    (
        "AWS provides cloud services.",
        {"entities": [(0, 3, "CLOUD_PLATFORM")]}
    ),
    (
        "PostgreSQL is a database.",
        {"entities": [(0, 10, "DATABASE")]}
    ),
    (
        "Power BI is used for visualization.",
        {"entities": [(0, 8, "TOOL")]}
    )
]

print("Training examples:", len(TRAIN_DATA))
```

```python
from spacy.training import Example

ner = nlp.get_pipe("ner")

labels = [
    "SKILL",
    "TOOL",
    "TECHNOLOGY",
    "DATABASE",
    "CLOUD_PLATFORM"
]

for label in labels:
    ner.add_label(label)

print("Custom labels added successfully")
```

```python
custom_entities = {
    "Python": "SKILL",
    "AWS": "CLOUD_PLATFORM",
    "PostgreSQL": "DATABASE",
    "Power BI": "TOOL"
}

for entity, label in custom_entities.items():
    print(entity, "->", label)
```
