```python
import re

def clean_text(text):
    text = str(text)

    # Remove URLs
    text = re.sub(r"http\S+", "", text)

    # Remove HTML tags
    text = re.sub(r"<.*?>", "", text)

    # Remove email addresses
    text = re.sub(r"\S+@\S+", "", text)

    # Remove special characters
    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text)

    return text.strip().lower()
```

```python
df["clean_description"] = df["description"].apply(clean_text)

print("Cleaning applied successfully!")
```

```python
master_skills = [
    "Python",
    "SQL",
    "Java",
    "C++",
    "Power BI",
    "Tableau",
    "Excel",
    "AWS",
    "Azure",
    "GCP",
    "TensorFlow",
    "PyTorch",
    "Pandas",
    "NumPy",
    "Scikit-learn",
    "Spark",
    "Hadoop",
    "Docker",
    "Kubernetes",
    "Git"
]

print("Total master skills:", len(master_skills))
print(master_skills)
```

```python
skill_categories = {
    "Programming": [
        "Python", "Java", "C++", "Git"
    ],

    "Database": [
        "SQL"
    ],

    "Data Analytics": [
        "Power BI", "Tableau", "Excel", "Pandas", "NumPy"
    ],

    "Machine Learning": [
        "TensorFlow", "PyTorch", "Scikit-learn", "Spark", "Hadoop"
    ],

    "Cloud": [
        "AWS", "Azure", "GCP", "Docker", "Kubernetes"
    ]
}

print(skill_categories)
```

```python
skill_aliases = {
    "Python": "python3",
    "Excel": "ms excel",
    "PostgreSQL": "postgres, postgresql",
    "Power BI": "powerbi, power-bi",
    "Scikit-learn": "scikit learn"
}

print(skill_aliases)
```

```python
skill_taxonomy = []

for category, skills in skill_categories.items():
    for skill in skills:
        alias = skill_aliases.get(skill, "")

        skill_taxonomy.append({
            "skill": skill,
            "category": category,
            "aliases": alias
        })

skill_taxonomy_df = pd.DataFrame(skill_taxonomy)

skill_taxonomy_df.to_csv("skill_taxonomy.csv", index=False)

print("Skill taxonomy created successfully!")
print(skill_taxonomy_df)
```
