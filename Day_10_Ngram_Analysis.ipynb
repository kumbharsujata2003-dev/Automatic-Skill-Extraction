```python
import pandas as pd
```

```python
df = pd.read_csv("raw_jobs.csv")
```

```python
df["clean_description"] = df["description"].fillna("").astype(str)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    ngram_range=(1,3),
    stop_words="english"
)
```

```python
X = vectorizer.fit_transform(df["clean_description"])
```

```python
terms = vectorizer.get_feature_names_out()
scores = X.sum(axis=0).A1

ngram_keywords = sorted(
    zip(terms, scores),
    key=lambda x: x[1],
    reverse=True
)

ngram_keywords[:20]
```

#Conclusion:
N-gram analysis helps discover multi-word skills and phrases from job descriptions.
