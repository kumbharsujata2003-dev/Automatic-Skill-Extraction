```python
%%writefile utils/skill_extractor.py

def extract_skills(text):
    skills = ["Python", "SQL", "Power BI", "Excel"]

    detected_skills = []

    for skill in skills:
        if skill.lower() in text.lower():
            detected_skills.append(skill)

    return detected_skills
```

```python
%%writefile utils/category_mapping.py

def get_category(skill):
    categories = {
        "Python": "Programming",
        "SQL": "Database",
        "Power BI": "BI",
        "Excel": "Analytics"
    }

    return categories.get(skill, "Other")
```

```python
%%writefile app.py

import streamlit as st
from utils.skill_extractor import extract_skills
from utils.category_mapping import get_category

st.title("Automatic Skill Extraction from Job Descriptions")

st.write("Paste a job description below.")

job_description = st.text_area(
    "Paste Job Description",
    placeholder="Looking for Data Analyst with SQL, Python, Power BI and Excel experience."
)

if st.button("Extract Skills"):
    detected_skills = extract_skills(job_description)

    st.subheader("Detected Skills")

    if detected_skills:
        st.write(", ".join(detected_skills))

        st.subheader("Categories")

        for skill in detected_skills:
            category = get_category(skill)
            st.write(f"{category} → {skill}")
    else:
        st.write("No skills detected.")
```
